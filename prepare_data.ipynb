```python
import pandas as pd
import json
```

```python
save_path = '/NS/llm-pretraining/work/afkhan/HubbleSuite/open-unlearning/Hubble_Data/'
```

# Yago Biographies

```python
# Login using e.g. `huggingface-cli login` to access this dataset
splits = {'train': 'train.csv', 'test': 'test.csv'}
df_train = pd.read_csv("hf://datasets/allegrolab/biographies_yago/" + splits["train"])
df_test = pd.read_csv("hf://datasets/allegrolab/biographies_yago/" + splits["test"])
```

```python
train_save_path = save_path + 'biographies_yago_train.jsonl'

train_jsonl_data = []

for index, row in df_train.iterrows():
    row_as_dict = row.to_dict()
    line = {
        'text': row_as_dict['text']
    }
    train_jsonl_data.append(line)

with open(train_save_path, 'w') as f:
    for item in train_jsonl_data:
        f.write(json.dumps(item) + '\n')
```

```python
dup_counts = list(df_train['duplicates'].unique())

for dup in dup_counts:
    train_save_path = save_path + f'biographies_yago_train_dup_{dup}.jsonl'
    df_train_subset = df_train[df_train['duplicates'] == dup]
    print(f"Processing {len(df_train_subset)} rows for duplicate {dup}")
    train_jsonl_data = []
    for index, row in df_train_subset.iterrows():
        row_as_dict = row.to_dict()
        line = {
            'text': row_as_dict['text']
        }
        train_jsonl_data.append(line)

    with open(train_save_path, 'w') as f:
        for item in train_jsonl_data:
            f.write(json.dumps(item) + '\n')
```

```python
test_save_path = save_path + 'biographies_yago_test.jsonl'

test_jsonl_data = []

for index, row in df_test.iterrows():
    row_as_dict = row.to_dict()
    line = {
        'text': row_as_dict['text']
    }
    test_jsonl_data.append(line)

with open(test_save_path, 'w') as f:
    for item in test_jsonl_data:
        f.write(json.dumps(item) + '\n')
```
